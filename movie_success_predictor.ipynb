```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('data/TMDB_movie_dataset_v11.csv')
print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
# Check how much of the data is actually usable
print(df['status'].value_counts())
print("Movies with revenue > 0:", (df['revenue'] > 0).sum())
print("Movies with budget > 0:", (df['budget'] > 0).sum())
print("Movies with BOTH budget > 0 and revenue > 0:", ((df['budget'] > 0) & (df['revenue'] > 0)).sum())
```

```python
df_clean = df[(df['budget'] > 0) & (df['revenue'] > 0) & (df['status'] == 'Released')].copy()
print(df_clean.shape)
df_clean.head()
```

```python
# Check missing values in our cleaned data
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])

# Check vote_count distribution — low vote counts mean unreliable ratings
print(df_clean['vote_count'].describe())

# Check runtime for zero/missing values
print("Movies with runtime = 0:", (df_clean['runtime'] == 0).sum())
```

```python
df_clean = df_clean.drop(columns=[
    'vote_average', 'vote_count', 'popularity',
    'backdrop_path', 'homepage', 'poster_path', 'imdb_id',
    'overview', 'tagline', 'adult', 'original_title', 'id', 'status', 'keywords'
])

df_clean = df_clean[df_clean['genres'].notna()]
df_clean = df_clean[df_clean['runtime'] > 0]

print(df_clean.shape)
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
# Drop rows missing release_date (small number, essential feature)
df_clean = df_clean[df_clean['release_date'].notna()]

# Fill remaining categorical missing values with 'Unknown' instead of dropping more rows
df_clean['production_companies'] = df_clean['production_companies'].fillna('Unknown')
df_clean['production_countries'] = df_clean['production_countries'].fillna('Unknown')
df_clean['spoken_languages'] = df_clean['spoken_languages'].fillna('Unknown')

print(df_clean.shape)
print(df_clean.isnull().sum().sum())  # should be 0 now
```

```python
df_clean['release_date'] = pd.to_datetime(df_clean['release_date'])
df_clean['release_year'] = df_clean['release_date'].dt.year
df_clean['release_month'] = df_clean['release_date'].dt.month

df_clean[['release_date', 'release_year', 'release_month']].head()
```

```python
# Split the comma-separated genre string into a list per row
df_clean['genres_list'] = df_clean['genres'].apply(lambda x: [g.strip() for g in x.split(',')])

# Create one binary column per genre (1 if movie has that genre, 0 if not)
from sklearn.preprocessing import MultiLabelBinarizer

mlb = MultiLabelBinarizer()
genre_encoded = pd.DataFrame(mlb.fit_transform(df_clean['genres_list']), columns=mlb.classes_, index=df_clean.index)

print(genre_encoded.shape)
genre_encoded.head()
```

```python
df_clean['num_production_companies'] = df_clean['production_companies'].apply(
    lambda x: len(x.split(',')) if x != 'Unknown' else 0
)
df_clean['num_production_countries'] = df_clean['production_countries'].apply(
    lambda x: len(x.split(',')) if x != 'Unknown' else 0
)
df_clean['num_spoken_languages'] = df_clean['spoken_languages'].apply(
    lambda x: len(x.split(',')) if x != 'Unknown' else 0
)

df_clean[['num_production_companies', 'num_production_countries', 'num_spoken_languages']].describe()
```

```python
# Combine numeric features + genre one-hot columns
feature_cols = ['budget', 'runtime', 'release_year', 'release_month',
                 'num_production_companies', 'num_production_countries', 'num_spoken_languages']

X = pd.concat([df_clean[feature_cols], genre_encoded], axis=1)
y = df_clean['revenue']

print(X.shape)
print(y.describe())
```

```python
import numpy as np

y_log = np.log1p(y)

print(y_log.describe())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y_log, test_size=0.2, random_state=42
)

print(X_train.shape, X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

y_pred = lr_model.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("RMSE (log scale):", rmse)
print("R²:", r2)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)

rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

print("RMSE (log scale):", rmse_rf)
print("R²:", r2_rf)
```

```python
importances = pd.Series(rf_model.feature_importances_, index=X.columns).sort_values(ascending=False)
print(importances.head(10))
```

```python
import joblib

joblib.dump(rf_model, 'movie_revenue_model.pkl')
joblib.dump(list(X.columns), 'model_features.pkl')  # save exact column order/names for the API

print("Model and feature list saved!")
```
